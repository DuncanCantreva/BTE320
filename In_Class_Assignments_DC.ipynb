```python
num_1 = float(input("Number 1: "))
num_2 = float(input("Number 2: "))
Symbol = input("Mathematical Symbol: ")

if Symbol == "+":
  print(num_1 + num_2)
elif Symbol == "-":
  print(num_1 - num_2)
elif Symbol == "/":
  print(num_1 / num_2)
elif Symbol == "*":
  print(num_1 * num_2)
elif Symbol == "**":
  print(num_1 ** num_2)
elif Symbol == "%":
  print(num_1 % num_2)
else:
  print("Invalid Operation")
```

```python
start = input("Do you want to play the game? (y/n)")
import random
while start == "y":
  computer_move = random.choice(["rock", "paper", "scissors"])
  your_move = input("Your Move: ")
  print(f"Computers Move: {computer_move}")

  if your_move == computer_move:
    print("Draw")

  elif your_move == "rock" and computer_move == "scissors":
    print("You Win!")
  elif your_move == "scissors" and computer_move == "paper":
    print("You Win!")
  elif your_move == "paper" and computer_move == "rock":
    print("You Win!")
  else:
    print("You Lose!")

  start = input("Do you want to play again? (y/n)")
```

```python
def getValues(n):
    items = []
    for i in range(n):
        items.append(float(input("Enter a value: ")))
    return items

def calcAverage(alist):
    assert isinstance(alist, list), "'alist' must be a list"

    if len(alist) < 3:
        raise ValueError("The list 'alist' must have at least 3 items to calculate this average")

    subset = alist[2:]
    return sum(subset) / len(subset)

def main():
    try:
        user_input = input("Enter the number of items: ")
        n = int(user_input)

        alist = getValues(n)
        result = calcAverage(alist)

        print("The average of the items (excluding the first two) is:", result)

    except ValueError as e:
        print("Error:", e)
    except ZeroDivisionError:
        print("Error: Could not divide by zero.")

main()
```
